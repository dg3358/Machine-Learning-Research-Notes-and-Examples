```python
import numpy as np
from scipy.stats import binom
```

```python
## Grid of possible values the proportion of water can take in the globe tossing example
p_grid = np.arange(0, 1.001, 0.001)
## Prior distribution for the proportion of water in the globe tossing example
prior = np.repeat(1, 1001)
## Likelihood of observing 6 water and 3 land in 9 tosses of the globe
likelihood = binom.pmf(k=6, n=9, p=p_grid)
## Posterior distribution for the proportion of water in the globe tossing example
posterior = likelihood * prior / ((likelihood * prior).sum())

np.random.seed(100)
## Sample of proportion of water p from posterior
samples = np.random.choice(p_grid, size=10000, replace=True, p=posterior)
```

### 3E1

Proportion of posterior probability below p = 0.2

```python
## From posterior itself
print("P(p < 0.2) from posterior: ", posterior[:np.where(p_grid == 0.2)[0][0]].sum())
## from sample
print("P(p < 0.2) from sample: ", np.mean(samples < 0.2))
```

### 3E2

Proportion of posterior probability above p = 0.8

```python
## From posterior itself
print("P(p > 0.8) from posterior: ", posterior[np.where(p_grid == 0.8)[0][0]:].sum())
## from sample
print("P(p > 0.8) from sample: ", np.mean(samples > 0.8))
```
