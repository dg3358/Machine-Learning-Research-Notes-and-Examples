## 1.1 - Statistical Golems

- Scientific/Statistical models are constructs engineered for a particular purpose, neither fully true or false
- Learning simply about the pre-built models for statistical testing without understanding the inner workings is akin to learning bridge building without understanding basic physics
- What is neeeded is a unified theory of golem engineering, or a set of principles for designing, building, and refining special purpose statistical procedures.

## 1.2 - Statistical Rethinking

- **Myth**: THe proper objective of statistical inference is to test null hypotheses, a belief derived from Karl Popper's falsifiability hypothesis.
    - 2 reasons this isn't true:
        1. Hypotheses are not models, and the relationships among hypotheses and different models is complex. Any given statistical model can correspond to more than one hypothesis, and visa versa.
            - Example: When scientists were debating the "neutral" theory of evolution, they debated the following - $H_0:$ "Evolution is Neutral" vs. $H_1:$ "Selection Matters". However, there is the additional question as to whether natural selection will always favor certain alleles or selection fluctuates through time. The statistical model of frequency of alleles that would support the null hypothesis of Neutral Evolution with frequency in equilibrium is the same as the one that would support "Fluctuating Natural Selection".     
        2. Data may be flawed, and hypotheses are often not binary yes/no but instead continuous - hence, strict falsification is borderline impossible.
            - Example of flawed data: In 2011, a team of physicists thought they observed a faster than light particle, but it was really an observation error due to a poorly attached cable.
            - Example of continuous hyothesis: $H_0$: All swans are white (binary) vs. $H_0$: 80\% of swans are white (continuous)

## 1.3 - Tools for Golem Engineering

- 3 primrary purposes for model usage:
    1. Designing Inquiry
    2. Extracting Information from Data
    3. Making Predictions
- Concepts to help with this:
    1. Bayesian Data Analysis
        - Counting of the number of ways data could happen according to our assumptions
        - A purely logical way of processing the data to produce inference, once the statistical model has been specified.
        - Treats uncertainty as a product of the INFORMATION provided to the mode; if we had more information, we could exactly predict everything
            - Contrasted with the Frequentist approach, which treats uncertainty as a product of SAMPLING, or variations in repeat measurements. This implies that there is no such thing as parameter uncertainty
    2. Model Comparison and Prediction
        - 2 tools for this cause: Cross-Validation and Information Criteria
        - Comparing model predictive accuracy leads you to to the counter-intuitive conclusion that simpler models are often better, due to the curse of over-fitting
    3. Multi-level Models [also knows as Hierarchical Models, Random Effects Models, and Mixed Effects Models]
        - Model parameters themselves can be subject to uncertainty; a model that explicitly allows for this may capture data better
        - **Partial Pooling**: An antidote to overfitting risk that pools information across units to produce better estimates for all units; useful in 4 specific situations:
            - Adjusting estimates for repeat sampling (to account for correlation across theses samples)
            - Adjust estimates for imbalance in sampling (accounting for greater volatlity in ertain groups)
            - To study variation (evaluating intra-class volatlity)
            - To avoid averaging (prevent from the manufacturing of false confidence in a certain group of estimates)
    4. Graphical Causal Models
        - Since statistical models only measure _associations_, and purely evaluating prediction may mislead the modeler as to the true casual relationships between variables, we need something outside the data as a framework from interpreting model results
        - A Graphical Causal Model called a directed acyclic graph can meet this challenge by allowing us to incorporate intuition into the process of deducing which models provide valid causal inferences
